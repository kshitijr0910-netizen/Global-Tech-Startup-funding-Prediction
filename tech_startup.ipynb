1. Startup Count by Domain — Bar Chart

Shows which technology domains have the most startups

```python
import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Domain",
    order=df["Domain"].value_counts().index
)

plt.title("Number of Startups by Domain")
plt.xlabel("Number of Startups")
plt.ylabel("Domain")
plt.show()
```

2. Total Funding by Funding Stage — Box Plot

Useful for comparing funding distributions across stages.

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Funding_Stage",
    y="Total_Funding_USD_Millions"
)

plt.title("Funding Distribution by Funding Stage")
plt.xlabel("Funding Stage")
plt.ylabel("Total Funding (USD Millions)")
plt.xticks(rotation=45)
plt.show()
```

3. Funding vs Valuation — Scatter Plot

One of the most useful visualizations in this dataset.

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Total_Funding_USD_Millions",
    y="Valuation_USD_Millions",
    hue="Funding_Stage",
    alpha=0.5
)

plt.title("Total Funding vs Startup Valuation")
plt.xlabel("Total Funding (USD Millions)")
plt.ylabel("Valuation (USD Millions)")
plt.show()
```

4.Top 10 Domains by Total Funding

```python
top_domain = (df.groupby('Domain')['Total_Funding_USD_Millions']
                .sum()
                .sort_values(ascending=False)
                .head(10))

plt.figure(figsize=(10,6))
sns.barplot(x=top_domain.values, y=top_domain.index)
plt.title('Top 10 Domains by Total Funding')
plt.xlabel('Total Funding (USD Millions)')
plt.ylabel('Domain')
plt.show()
```

5. Top 10 Cities by Number of Startups

```python
top_cities = df['City'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_cities.values, y=top_cities.index)
plt.title('Top 10 Startup Cities')
plt.xlabel('Number of Startups')
plt.ylabel('City')
plt.show()
```

6. Average Valuation by Funding Stage

```python
avg_valuation = (df.groupby('Funding_Stage')['Valuation_USD_Millions']
                   .mean()
                   .sort_values(ascending=False))

plt.figure(figsize=(10,6))
sns.barplot(x=avg_valuation.index, y=avg_valuation.values)
plt.title('Average Valuation by Funding Stage')
plt.xlabel('Funding Stage')
plt.ylabel('Average Valuation (USD Millions)')
plt.xticks(rotation=45)
plt.show()
```

7.Revenue vs valuation

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    data=df,
    x='Revenue_ARR_Millions',
    y='Valuation_USD_Millions',
    hue='Funding_Stage'
)
plt.title('Revenue vs Valuation')
plt.xlabel('Revenue / ARR (USD Millions)')
plt.ylabel('Valuation (USD Millions)')
plt.show()
```

8.Layoff by domain

```python
layoffs = (df.groupby('Domain')['Layoffs_2024_2025']
             .sum()
             .sort_values(ascending=False)
             .head(10))

plt.figure(figsize=(10,6))
sns.barplot(x=layoffs.values, y=layoffs.index)
plt.title('Top 10 Domains by Layoffs')
plt.xlabel('Total Layoffs')
plt.ylabel('Domain')
plt.show()
```

9.Runway by Funding stage

```python
plt.figure(figsize=(10,6))
sns.boxplot(
    data=df,
    x='Funding_Stage',
    y='Runway_Months_2024'
)
plt.title('Runway by Funding Stage')
plt.xlabel('Funding Stage')
plt.ylabel('Runway (Months)')
plt.xticks(rotation=45)
plt.show()
```

10.Numerical correlation Heatmap

```python
numeric_cols = [
    'Total_Funding_USD_Millions',
    'Valuation_USD_Millions',
    'Revenue_ARR_Millions',
    'Runway_Months_2024',
    'Peak_Headcount_2023',
    'Layoffs_2024_2025',
    'Current_Headcount_2026'
]

plt.figure(figsize=(10,7))
sns.heatmap(
    df[numeric_cols].corr(),
    annot=True,
    fmt='.2f',
    cmap='coolwarm'
)
plt.title('Correlation Between Startup Metrics')
plt.show()
```

```python
y=df['Total_Funding_USD_Millions']
X=df[['Domain', 'Country', 'City', 'Funding_Stage', 'Valuation_USD_Millions','Runway_Months_2024',
 'Peak_Headcount_2023', 'Layoffs_2024_2025','Current_Headcount_2026']]
```

```python
 cat_cols = ['Domain', 'Country', 'City', 'Funding_Stage']

 num_cols = ['Valuation_USD_Millions','Runway_Months_2024', 'Peak_Headcount_2023', 'Layoffs_2024_2025',
       'Current_Headcount_2026']
```

```python
df.isnull().sum()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,MinMaxScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num",MinMaxScaler(),num_cols),
        ("cat",OneHotEncoder(),cat_cols),
    ]
)
```

```python
from sklearn .pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

model = Pipeline(
    steps = [
        ("preprocessing",preprocessor),
        ("regression",RandomForestRegressor())
    ]
)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred
```

```python
print(model.score(X_train,y_train))
print(model.score(X_test,y_test))
```
